## Question 1 : What is the distribution of the machines according to their CPU capacity ?

#### Load machine events CSV as RDD

```python
from pyspark import SparkContext
from pyspark.mllib.stat import Statistics

sc = SparkContext.getOrCreate()
sc.setLogLevel("ERROR")

raw_rdd = sc.textFile("../spark-data/machine_events/*")
raw_rdd.take(3)
```

#### Remove header:

```python
# header = raw_rdd.first()
# rdd = raw_rdd.filter(lambda line: line != header)
```

#### Parse lines

```python
def parse_machine_event(line):
    fields = line.split(",")

    # Must have exactly 6 fields for machine_events
    if len(fields) < 6:
        return None

    time = fields[0]
    machine_id = fields[1]
    cpu = fields[4]

    # CPUs is NOT mandatory
    if time == "" or machine_id == "" or cpu == "":
        return None

    try:
        return (int(machine_id), (int(time), float(cpu)))
    except:
        return None

parsed_rdd = raw_rdd.map(parse_machine_event)

clean_rdd = parsed_rdd.filter(lambda x: x is not None)
```

#### Keep the latest CPU capacity per machine

```python
latest_cpu_per_machine = clean_rdd.reduceByKey(
    lambda a, b: a if a[0] > b[0] else b
)
cpu_capacity_rdd = latest_cpu_per_machine.map(lambda x: x[1][1])
```

```python
cpu_distribution = (
    cpu_capacity_rdd
    .map(lambda cpu: (cpu, 1))
    .reduceByKey(lambda a, b: a + b)
    .sortByKey()
)

cpu_distribution.collect()
```

#### Plot the CPU capacity histogram

```python
data = cpu_distribution.collect()

cpu_values = [x[0] for x in data]
counts = [x[1] for x in data]

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(cpu_values, counts, width=0.1)

plt.xlabel("CPU Capacity (normalized)")
plt.ylabel("Number of Machines")
plt.title("Distribution of Machines by CPU Capacity")

plt.xticks(cpu_values)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```

The figure shows the distribution of machines according to their normalized CPU capacity. Most machines have a CPU capacity of 0.5, while fewer machines provide lower (0.25) or higher (1.0) CPU capacities. This confirms that the cluster is heterogeneous, with a dominant machine configuration and a smaller number of more specialized machines.

## Question 3 : Is there a class of machines, according to their CPU, that stands out with a higher maintenance rate, as compared to other classes ?

The goal of this analysis is to determine whether some classes of machines, grouped according to their CPU capacity, experience a higher maintenance rate than others.

In the context of this dataset, maintenance is inferred from machine REMOVE events, which indicate that a machine was taken offline and later reintroduced into the cluster.

We use the **machine_events table**. Each machine may appear multiple times in the dataset due to configuration changes or maintenance events.

#### Step 1 : Parse machine events (schema-aware)
We parse the machine_events files according to the official schema and extract the fields.

Rows with missing mandatory fields are discarded.

```python
def parse_machine_event_full(line):
    fields = line.split(",")

    if len(fields) < 6:
        return None

    time = fields[0]
    machine_id = fields[1]
    event_type = fields[2]
    cpu = fields[4]

    if time == "" or machine_id == "" or event_type == "":
        return None

    try:
        cpu_val = float(cpu) if cpu != "" else None
        return (int(machine_id), (int(time), int(event_type), cpu_val))
    except:
        return None

events_rdd = (
    rdd
    .map(parse_machine_event_full)
    .filter(lambda x: x is not None)
)
```

#### Step 2 : Get latest CPU capacity per machine
Since a machine can appear multiple times, we associate each machine with its latest known CPU capacity (maximum timestamp).

This gives a single, consistent CPU value per machine.

```python
machine_cpu = (
    events_rdd
    .filter(lambda x: x[1][2] is not None)  # cpu exists
    .map(lambda x: (x[0], (x[1][0], x[1][2])))
    .reduceByKey(lambda a, b: a if a[0] > b[0] else b)
    .mapValues(lambda x: x[1])
)
```

#### Step 3 : Count REMOVE events per machine
In the dataset, a **REMOVE event (event_type = 1)** indicates that a machine was taken offline.

We count the number of REMOVE events per machine and use this count as a proxy for maintenance frequency.

```python
REMOVE_EVENT = 1

remove_events = (
    events_rdd
    .filter(lambda x: x[1][1] == REMOVE_EVENT)
    .map(lambda x: (x[0], 1))
    .reduceByKey(lambda a, b: a + b)
)
```

#### Step 4 : Join CPU capacity with REMOVE counts

```python
machine_maintenance = machine_cpu.leftOuterJoin(remove_events)

machine_maintenance = machine_maintenance.mapValues(
    lambda x: (x[0], x[1] if x[1] is not None else 0)
)
```

#### Step 5 : Aggregate by CPU class and compute maintenance rate
From Question 1, we observed that machines have only three CPU capacities:

* 0.25 → Low CPU
* 0.5 → Medium CPU
* 1.0 → High CPU

These values are used to define CPU classes.

For each CPU class, we compute the **maintenance rate** as follows:

**Maintenance Rate = Total REMOVE events / Number of machines in the class**


This metric represents the **average number of maintenance events per machine** within each CPU class.

```python
cpu_class_stats = (
    machine_maintenance
    .map(lambda x: (x[1][0], (x[1][1], 1)))  # cpu → (removals, machines)
    .reduceByKey(lambda a, b: (a[0] + b[0], a[1] + b[1]))
)

maintenance_rate = cpu_class_stats.mapValues(
    lambda x: x[0] / x[1]   # removals per machine
)

for cpu, rate in maintenance_rate.collect():
    print(f"CPU {cpu}: maintenance rate = {round(rate, 3)}")
```

#### Plot the result

```python
import matplotlib.pyplot as plt

data = [(0.5, 0.7056), (0.25, 0.8413), (1.0, 0.7820)]
cpu = [x[0] for x in data]
rate = [x[1] for x in data]

plt.figure(figsize=(7,5))
plt.bar(cpu, rate, width=0.15)
plt.xlabel("CPU Capacity")
plt.ylabel("Maintenance Rate (REMOVE events per machine)")
plt.title("Maintenance Rate by Machine CPU Capacity")
plt.xticks(cpu)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

We computed the maintenance rate of machines by counting the number of REMOVE events per machine and aggregating the results by CPU capacity class. The results show that machines with low CPU capacity (0.25) have the highest maintenance rate, followed by high-capacity machines (1.0), while medium-capacity machines (0.5) are the most stable. This suggests that older or less powerful machines may be more prone to failures, while high-end machines may undergo more frequent maintenance due to higher utilization or upgrades.

#### Results

Maintenance rates vary across CPU classes:
- **Low-CPU machines** have the highest maintenance rate
- **Medium-CPU machines** are the most stable
- **High-CPU machines** show a moderate maintenance rate

These differences are illustrated using a bar chart.

### Interpretation

Low-capacity machines may correspond to older or less reliable hardware.  
Medium-capacity machines appear to offer a good balance between performance and reliability.  
High-capacity machines may require more maintenance due to higher utilization or upgrades.

### Conclusion

Machine maintenance behavior differs across CPU classes, with **low-CPU machines standing out as the least stable**.


## Question 5 : Would you qualify the percentage of jobs/tasks that got killed or evicted as important?

In a large-scale shared cluster, not all jobs and tasks complete successfully. Some may be killed due to user intervention or system decisions, while others may be evicted to free resources for higher-priority workloads.

The goal of this analysis is to quantify the proportion of jobs and tasks that were killed or evicted, and to assess whether this proportion can be considered significant.

### Job-Level Analysis
We first analyze the problem at the job level using the **job_events** table.

Each job goes through multiple state changes during its lifetime. A job is considered killed or evicted if it experiences at least one **KILL or EVICT** event during execution.

#### Step 1 : Parse job events
We start by parsing the `job_events` table.
Each record describes a state change in the lifecycle of a job.

Malformed records and records with missing mandatory fields are discarded.
The result is an RDD mapping each job to its corresponding event types.

```python
def parse_job_event(line):
    fields = line.split(",")

    if len(fields) < 4:
        return None

    job_id = fields[2]
    event_type = fields[3]

    if job_id == "" or event_type == "":
        return None

    try:
        return (int(job_id), int(event_type))
    except:
        return None
```

```python
job_events_rdd = (
    sc.textFile("../spark-data/job_events/*.csv.gz")
    .map(parse_job_event)
    .filter(lambda x: x is not None)
)

job_events_rdd.take(5)
```

#### Step 2 : Identifying killed or evicted jobs
According to the Google Cluster documentation:
- event type `2` corresponds to EVICT
- event type `5` corresponds to KILL

A job is considered *killed or evicted* if it experienced **at least one** such event
during its lifetime.

We compute:
- the total number of distinct jobs
- the number of jobs that were killed or evicted at least once

```python
EVICT = 2
KILL = 5

# Total number of jobs
total_jobs = job_events_rdd.map(lambda x: x[0]).distinct().count()
print(f"Total number of jobs: {total_jobs}")

# Jobs that were killed or evicted at least once
killed_or_evicted_jobs = (
    job_events_rdd
    .filter(lambda x: x[1] in (KILL, EVICT))
    .map(lambda x: x[0])
    .distinct()
    .count()
)
print(f"Number of jobs killed or evicted: {killed_or_evicted_jobs}")
```

#### Step 3 : Percentage (jobs)
We compute the percentage of jobs that were killed or evicted by dividing
the number of affected jobs by the total number of jobs.

This metric provides a high-level view of job-level failures or preemptions
in the cluster.

```python
job_failure_rate = killed_or_evicted_jobs / total_jobs * 100
print(f"Percentage of jobs killed or evicted: {round(job_failure_rate, 2)}%")
```

### Results Analysis
Approximately **37.63% of jobs** experienced at least one KILL or EVICT event during their lifetime.
This indicates that more than one third of submitted jobs did not run to completion without interruption.

Such a proportion suggests that job-level preemption and termination are **not rare** in the cluster.
This behavior is consistent with a shared production environment where jobs may be killed due to
resource constraints, scheduling policies, or user intervention.

### Task-Level Analysis
We then repeat the same analysis at the task level using the task_events table.
A task is considered killed or evicted if it experiences at least one KILL or EVICT event.

#### Step 1 : Parse task events
We now repeat the analysis at the task level using the `task_events` table.

Each task is uniquely identified by the pair `(job_id, task_index)`.

As before, malformed records and records with missing mandatory fields are removed.

```python
def parse_task_event(line):
    fields = line.split(",")

    if len(fields) < 6:
        return None

    job_id = fields[2]
    task_index = fields[3]
    event_type = fields[5]

    if job_id == "" or task_index == "" or event_type == "":
        return None

    try:
        return ((int(job_id), int(task_index)), int(event_type))
    except:
        return None
```

```python
task_events_rdd = (
    sc.textFile("../spark-data/task_events/*.csv.gz")
    .map(parse_task_event)
    .filter(lambda x: x is not None)
)
```

#### Step 2 : Identifying killed or evicted tasks
Using the same event type codes as for jobs (EVICT = 2, KILL = 5),
we identify tasks that were killed or evicted at least once.

We compute:
- the total number of distinct tasks
- the number of tasks that experienced a KILL or EVICT event

```python
# Total number of tasks
total_tasks = task_events_rdd.map(lambda x: x[0]).distinct().count()
print(f"Total number of tasks: {total_tasks}")

# Tasks that were killed or evicted at least once
killed_or_evicted_tasks = (
    task_events_rdd
    .filter(lambda x: x[1] in (KILL, EVICT))
    .map(lambda x: x[0])
    .distinct()
    .count()
)
print(f"Number of tasks killed or evicted: {killed_or_evicted_tasks}")
```

#### Step 3 : Computing the percentage of affected tasks
We compute the percentage of tasks that were killed or evicted by dividing
the number of affected tasks by the total number of tasks.

Comparing task-level and job-level percentages helps us understand
whether failures are localized to individual tasks or impact entire jobs.

```python
task_failure_rate = killed_or_evicted_tasks / total_tasks * 100
print(f"Percentage of tasks killed or evicted: {round(task_failure_rate, 2)}%")
```

### Results Analysis
At the task level, **31.6% of tasks** were killed or evicted at least once.
This percentage is slightly lower than the job-level rate.

This difference can be explained by the fact that a job may be marked as killed or evicted
even if only a subset of its tasks is affected.
Conversely, many tasks may complete successfully even when their parent job experiences
temporary disruptions.

Task evictions are therefore a common mechanism used by the scheduler to dynamically
reallocate resources without necessarily failing entire jobs.

#### Conclusion
The proportion of killed or evicted jobs and tasks is **significant**, showing that eviction and
preemption are common mechanisms used by the scheduler to manage resource contention in the cluster.

## Question 7 :  In general, do tasks from the same job run on the same machine? Comment on the observed locality strategy and its pros and cons.

The objective of this analysis is to determine whether tasks belonging to the same job
tend to run on the same machine, or whether they are spread across multiple machines.

This helps us understand the locality strategy used by the scheduler and its potential
advantages and drawbacks.

#### Step 1 : Parse task events (schema-aware)

```python
def parse_task_schedule(line):
    fields = line.split(",")

    if len(fields) < 6:
        return None

    job_id = fields[2]
    task_index = fields[3]
    machine_id = fields[4]
    event_type = fields[5]

    if job_id == "" or task_index == "" or machine_id == "" or event_type == "":
        return None

    try:
        return (int(job_id), int(task_index), int(machine_id), int(event_type))
    except:
        return None
```

```python
task_events_rdd = (
    sc.textFile("../spark-data/task_events/*.csv.gz")
    .map(parse_task_schedule)
    .filter(lambda x: x is not None)
)

task_events_rdd.take(5)
```

#### Step 2 : Keep only SCHEDULE events
We focus only on SCHEDULE events, which indicate when a task is assigned to a machine.
This allows us to analyze actual task placement decisions made by the scheduler.

```python
SCHEDULE = 1

scheduled_tasks = task_events_rdd.filter(lambda x: x[3] == SCHEDULE)
```

#### Step 3 : For each job, compute number of machines used
For each job, we compute the number of **distinct machines** on which its tasks were scheduled.
This metric directly reflects task locality:
- a value close to 1 indicates strong locality
- larger values indicate tasks spread across multiple machines

```python
job_machine_count = (
    scheduled_tasks
    .map(lambda x: (x[0], x[2]))  # job_id → machine_id
    .distinct()
    .map(lambda x: (x[0], 1))
    .reduceByKey(lambda a, b: a + b)
)

job_machine_count.take(5)
```

#### Step 4 : Compute average locality behavior
We compute the average number of machines used per job to capture the
overall locality behavior across the cluster.

```python
average_machines_per_job = job_machine_count.map(lambda x: x[1]).mean()
average_machines_per_job
print(f"Average locality behavior: {round(average_machines_per_job, 2)}")
```

### Interpretation

On average, tasks belonging to the same job are scheduled on approximately **31 different machines**.

This indicates that tasks from a single job are **widely distributed across the cluster**
rather than being executed on a single machine or a small set of machines.
Such behavior suggests a scheduling strategy that prioritizes **parallelism and load balancing**
over strict task locality.

### Conclusion

Overall, tasks from the same job do **not** generally run on the same machine.
Instead, the scheduler tends to spread tasks across many machines.

This strategy improves scalability and resource utilization, but may reduce data locality
and increase communication overhead.
The observed behavior highlights a trade-off between locality and efficient use of
cluster resources.

#### For each job, compute number of tasks

```python
job_task_count = (
    scheduled_tasks
    .map(lambda x: ((x[0], x[1]), 1))   # unique task
    .map(lambda x: (x[0][0], 1))
    .reduceByKey(lambda a, b: a + b)
)
job_locality = job_task_count.join(job_machine_count)

job_locality_ratio = job_locality.mapValues(
    lambda x: x[1] / x[0]   # machines / tasks
)

average_locality = job_locality_ratio.map(lambda x: x[1]).mean()
average_locality
```

The average locality ratio is 0.994, which is very close to 1. This indicates that, for most jobs, the number of machines used is almost equal to the number of tasks. In other words, tasks belonging to the same job are generally scheduled on different machines, and task co-location is rare.

This observation shows that the scheduler favors a distributed execution strategy, prioritizing parallelism and load balancing over strict task locality. By spreading tasks across many machines, the cluster can maximize throughput and reduce job execution time, especially for jobs with a large number of independent tasks.

However, this strategy also has drawbacks. Distributing tasks across multiple machines reduces data locality and may increase network communication and synchronization overhead. Despite these costs, the observed behavior is consistent with the design of large-scale cloud infrastructures, where resource utilization and fairness across users are more critical than keeping tasks from the same job on a single machine.

In general, tasks from the same job do not run on the same machine. The observed scheduling strategy prioritizes throughput and fairness over locality, which is consistent with the design of large-scale cloud infrastructures.

## Question 8 : Are tasks that request more resources the ones that consume more resources?

The objective of this analysis is to determine whether tasks that request more resources
(CPU and memory) are also the ones that consume more resources during execution.

To answer this question, we compare the resource requests specified at task scheduling
time with the actual resource usage measured during task execution.

#### Step 1 : Load & parse task resource requests (task_events)
We extract resource requests from the `task_events` table.
Each task specifies the amount of CPU and memory it requests at scheduling time.

Only **SCHEDULE events** are considered to ensure that each task contributes a single,
well-defined resource request.

```python
SCHEDULE = 1

def parse_task_requests(line):
    fields = line.split(",")

    if len(fields) < 11:
        return None

    job_id = fields[2]
    task_index = fields[3]
    event_type = fields[5]
    cpu_req = fields[9]
    mem_req = fields[10]

    if job_id == "" or task_index == "" or cpu_req == "" or mem_req == "":
        return None

    try:
        return (
            (int(job_id), int(task_index)),
            (int(event_type), float(cpu_req), float(mem_req))
        )
    except:
        return None
```

```python
task_requests = (
    sc.textFile("../spark-data/task_events/*.csv.gz")
    .map(parse_task_requests)
    .filter(lambda x: x is not None and x[1][0] == SCHEDULE)
    .map(lambda x: (x[0], (x[1][1], x[1][2])))  # (task → (cpu_req, mem_req))
    .reduceByKey(lambda a, b: a)   # keep one request per task
)
```

#### Step 2 : Load & parse task actual usage (task_usage)
We extract actual CPU and memory usage from the `task_usage` table.
Resource usage is reported over 5-minute time windows.

For each task, multiple usage records may exist.
These values will later be aggregated to compute average resource consumption.

```python
def parse_task_usage(line):
    fields = line.split(",")

    if len(fields) < 7:
        return None

    job_id = fields[2]
    task_index = fields[3]
    cpu_use = fields[5]
    mem_use = fields[6]

    if job_id == "" or task_index == "" or cpu_use == "" or mem_use == "":
        return None

    try:
        return (
            (int(job_id), int(task_index)),
            (float(cpu_use), float(mem_use))
        )
    except:
        return None
```

```python
task_usage = (
    sc.textFile("../spark-data/task_usage/*.csv.gz")
    .map(parse_task_usage)
    .filter(lambda x: x is not None)
)
```

#### Step 3 : Compute average usage per task
Since task usage is reported over multiple time windows, we compute the
**average CPU and memory usage per task**.

This provides a single representative consumption value for each task.

```python
task_avg_usage = (
    task_usage
    .mapValues(lambda x: (x[0], x[1], 1))
    .reduceByKey(lambda a, b: (
        a[0] + b[0],
        a[1] + b[1],
        a[2] + b[2]
    ))
    .mapValues(lambda x: (
        x[0] / x[2],   # average CPU usage
        x[1] / x[2]    # average memory usage
    ))
)
```

#### Step 4 : Join requests with usage
We join the task resource requests with the corresponding average usage
using the task identifier `(job_id, task_index)`.

```python
task_resources = task_requests.join(task_avg_usage)
```

#### Step 5 : Compute correlations
To quantify the relationship between requested and consumed resources,
we compute the **Pearson correlation coefficient** for CPU and memory.

Note:
- CPU requests are normalized values (0–1)
- CPU usage represents actual core usage

```python
cpu_pairs = task_resources.map(lambda x: (x[1][0][0], x[1][1][0]))
mem_pairs = task_resources.map(lambda x: (x[1][0][1], x[1][1][1]))

cpu_requested = cpu_pairs.map(lambda x: x[0])
cpu_used = cpu_pairs.map(lambda x: x[1])

cpu_correlation = Statistics.corr(cpu_requested, cpu_used)

mem_requested = mem_pairs.map(lambda x: x[0])
mem_used = mem_pairs.map(lambda x: x[1])

mem_correlation = Statistics.corr(mem_requested, mem_used)
```

```python
print("Correlation between requested and consumed resources:")
print(f"• CPU correlation   : {cpu_correlation:.3f}")
print(f"• Memory correlation: {mem_correlation:.3f}")
```

### Interpretation

The correlation between requested and consumed CPU is **moderate** (≈ 0.45).
This indicates that tasks requesting more CPU tend to consume more CPU on average,
but the relationship is not strong.

In contrast, the correlation between requested and consumed memory is **strong** (≈ 0.76).
This suggests that memory requests are generally good predictors of actual memory usage,
with less overestimation compared to CPU.

### Conclusion

Overall, tasks that request more resources tend to consume more resources,
but the relationship differs across resource types.

CPU requests appear to be more conservative and overestimated,
while memory requests are more closely aligned with actual usage.
This highlights differing resource estimation behaviors and suggests
potential optimization opportunities for CPU resource allocation.

In contrast, the correlation between requested memory and actual memory usage is strong (≈ 0.76). This shows that tasks requesting more memory generally consume more memory during execution. Memory requests therefore appear to be more accurate predictors of actual usage than CPU requests.

This difference can be explained by the nature of resources: CPU usage is highly dynamic and workload-dependent, while memory usage is more stable and often closely tied to the application’s data structures. As a result, users are better at estimating memory requirements than CPU needs.

Tasks that request more resources are more likely to consume more resources, but the strength of this relationship depends on the resource type. Memory requests are strongly correlated with actual consumption, whereas CPU requests show only a moderate correlation, highlighting inefficiencies and uncertainty in CPU resource estimation.

To capture periods of high resource pressure, we use the

## Question 11 : Do higher-priority tasks consume resources more efficiently than lower-priority tasks?

This question explores the relationship between **task priority** and **resource efficiency**,
which is not directly addressed in the provided questions.

While previous analyses focus on scheduling classes, evictions, or resource requests,
this question investigates whether higher-priority tasks make better use of the resources
they are allocated.

This analysis is original because it combines task priority, requested resources,
and actual resource usage to assess execution efficiency rather than scheduling outcomes.

### Step 1 : Parse task priority and CPU request
We extract task priority and requested CPU from the `task_events` table.
Only SCHEDULE events are considered to ensure each task contributes
a single, well-defined resource request.

```python
SCHEDULE = 1

def parse_task_priority_requests(line):
    fields = line.split(",")

    if len(fields) < 10:
        return None

    job_id = fields[2]
    task_index = fields[3]
    event_type = fields[5]
    priority = fields[8]
    cpu_req = fields[9]

    if job_id == "" or task_index == "" or priority == "" or cpu_req == "":
        return None

    try:
        return (
            (int(job_id), int(task_index)),
            (int(event_type), int(priority), float(cpu_req))
        )
    except:
        return None


task_priority_requests = (
    sc.textFile("../spark-data/task_events/*.csv.gz")
    .map(parse_task_priority_requests)
    .filter(lambda x: x is not None and x[1][0] == SCHEDULE)
    .map(lambda x: (x[0], (x[1][1], x[1][2])))  # (task → (priority, cpu_req))
    .reduceByKey(lambda a, b: a)
)
```

### Step 2 : Parse average CPU usage per task
We use the `task_usage` table to compute the average CPU consumption
for each task over its execution lifetime.

```python
def parse_task_cpu_usage(line):
    fields = line.split(",")

    if len(fields) < 6:
        return None

    job_id = fields[2]
    task_index = fields[3]
    cpu_use = fields[5]

    if job_id == "" or task_index == "" or cpu_use == "":
        return None

    try:
        return (
            (int(job_id), int(task_index)),
            (float(cpu_use), 1)
        )
    except:
        return None


task_cpu_usage = (
    sc.textFile("../spark-data/task_usage/*.csv.gz")
    .map(parse_task_cpu_usage)
    .filter(lambda x: x is not None)
    .reduceByKey(lambda a, b: (a[0] + b[0], a[1] + b[1]))
    .mapValues(lambda x: x[0] / x[1])  # avg CPU usage
)
```

### Step 3 : Computing CPU efficiency

CPU efficiency is defined as: **CPU efficiency = average CPU used / CPU requested**

This metric reflects how much of the allocated CPU a task actually consumes.

```python
task_efficiency = task_priority_requests.join(task_cpu_usage)

task_efficiency = task_efficiency.mapValues(
    lambda x: (
        x[0][0],                    # priority
        x[1] / x[0][1] if x[0][1] > 0 else 0.0  # efficiency
    )
)
```

### Step 4 : Aggregating efficiency by priority

We compute the average CPU efficiency for each priority level
to compare resource utilization across priorities.

```python
priority_efficiency = (
    task_efficiency
    .map(lambda x: (x[1][0], (x[1][1], 1)))
    .reduceByKey(lambda a, b: (a[0] + b[0], a[1] + b[1]))
    .mapValues(lambda x: x[0] / x[1])
    .sortByKey()
)

results = priority_efficiency.collect()
```

```python
print("Average CPU efficiency by task priority:")
for priority, efficiency in results:
    print(f"• Priority {priority}: average CPU efficiency = {efficiency:.3f}")
```

```python
import matplotlib.pyplot as plt

# Prepare data
priorities = [x[0] for x in results]
efficiencies = [x[1] for x in results]

plt.figure(figsize=(7, 4))
plt.bar(priorities, efficiencies)
plt.xlabel("Task Priority")
plt.ylabel("Average CPU Efficiency")
plt.title("Average CPU Efficiency by Task Priority")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

```python
from pyspark import SparkContext

sc = SparkContext.getOrCreate()

raw_rdd = sc.textFile("../spark-data/machine_events/")
raw_rdd.take(3)
```

2️⃣ Parse CSV lines correctly

```python
def parse_machine_event(line):
    fields = line.split(",")
    return (
        int(fields[1]),        # machine_id
        float(fields[4])       # cpu_capacity
    )

machine_rdd = raw_rdd.map(parse_machine_event)
```

```python
machine_cpu_rdd = machine_rdd.distinct()
```
